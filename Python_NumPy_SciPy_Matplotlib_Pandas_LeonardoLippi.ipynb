Exercício 19: Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

produto = np.dot(array1, array2)
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)
cosseno = produto / (norma1 * norma2)
angulo = np.arccos(cosseno)
print("Lista da primeira array: \n{}".format(array1))
print("\nLista da segunda array: \n{}".format(array2))
print("\nResultado do ângulo entre as duas arrays: \n{}".format(angulo))
```

Exercício 01: Calcule a integral da função f(x) = x 2 sin(x) no intervalo de 0 a π

```python
import numpy as np
from scipy import integrate

def f(x):
  return x**2 * np.sin(x)
resultado, erro = integrate.quad(f, 0, np.pi)
print("Resultado da integral: {}".format(resultado))
print("Estimativa de erro: {}".format(erro))
```

Exercício 02: Resolva a equação diferencial dy dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np

def diferencial(x,y):
  return x - y
solucao = solve_ivp(
    diferencial,
    [0, 10],
    [1],
    t_eval = np.linspace(0, 10, 100)
)
print("Tempo: \n{}".format(solucao.t))
print("\nSolução para Y: \n{}".format(solucao.y[0]))
```

Exercício 8: Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
limite_cima = y + 0.2
limite_baixo = y - 0.2

plt.plot(x, y, color='red', label='Seno(x)')
plt.fill_between(x, limite_baixo, limite_cima, color='blue')
plt.title("Gráfico de Seno com Intervalo de Confiança")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```
