Exercício 8: Verificação de Palíndromo

Neste exercício, o objetivo é mostrar se a frase ou palavra digitada é um palíndromo.

Começo criando a variável frase que vai receber a frase ou palavra digitada, depois faço a frase = frase.replace(" ", "") para remover os espaços na frase ou palavra.

Depois crio a variável invertida vazia, que vai ser responsável por armazenar a frase digitada porém ao contrário. Para isso vou utilizar o for com range começando pelo último caractere da frase ou palavra e indo até a primeira o len() vai contar o tamanho da frase ou palavra pelos seus caracteres o -1 um é para mostrar que vai começar pelo último índice e vai até o primeiro índice de trás para frente. A cada repetição, ele adiciona uma caractere encontra porém invertida.

Por fim, faço a comparação da frase invertida usando o if, if frasse == invertida, se a frase ou palavra forém iguais ela é um palíndromo, se não ele informa que ela não é.

```python
frase = input("Digite uma frase ou palavra: ")
frase = frase.replace(" ","")
invertida = ""

for f in range(len(frase)-1, -1, -1):
  invertida = invertida + frase[f]
  if frase == invertida:
    print("A frase é um PALÍNDROMO")
  else:
    print("A frase NÃO é um PALÍNDROMO")
```

Após isso crio o for dentro da lista, para percorrer a lista de dentro do for crio o if que vai ser responsavél por procurar o maior número, if n > maior o n é vai representar o número atual que está percorrendo pelo for e depois eu atualizo usando maior = n, faço a mesma coisa para o menor.

No final eu utilizo a função print() para exibir na tela o maior número e o menor número.

```python
numeros = [1,3,7,9,5,2,13]
maior = numeros[0]
menor = numeros[0]

for n in numeros:
  if n > maior:
    maior = n
  if n < menor:
    menor = n

print("O maior número é: {}".format(maior))
print("O menor número é: {}".format(menor))
```

Exercício 21: Busca Binária

Neste exercício, o objetivo era criar uma a busca binária para encontrar um elemento em uma lista ordenada.

Para isso começo criando uma lista chamada numeros, que já está em ordem crescente. Depois crio a variável busca que irá receber o número que o usuário deseja procurar.

Depois crio as variaveis inicio e fim para definir onde começa e termina a busca.

Utilizo o while para continuar procurando enquanto inicio for menor ou igual a fim. Dentro dele calculo o índice do meio usando (inicio + fim) // 2.

Depois utilizo o if para verificar se o número do meio é igual ao número buscado. Caso seja menor, avanço o inicio para procurar na parte da direita, e caso seja maior, diminuo o fim para procurar na parte da esquerda.Se o número for encontrado, utilizo o break para parar o laço.

Os prints serão utilizados junto do format para mostrar o número que foi digitado pelo usuário e sua posição, se não for encontrado ele irá mostrar que o número digitado pelo usuário não foi encontrado

```python
numeros = [2, 4, 8, 13, 17, 29, 35, 42]
busca = int(input("Digite o número inteiro que deseja procurar: "))
inicio = 0
fim = len(numeros) - 1

while inicio <= fim:
  meio = (inicio + fim) // 2

  if numeros[meio] == busca:
    print("O número {} foi encontrado na posição {}!".format(busca, meio))
    break
  elif numeros[meio] < busca:
    inicio = meio + 1
  else:
    fim = meio - 1
else:
  print("O número {} não está presente na lista...".format(busca))
```

Exercício 22: Criptografia de Cifra de César

Neste exercício, o objetivo era criar uma função para criptografar uma mensagem usando a cifra de César.

Primeiro começo criando a função cujo o nome é criptografar que vai receber mensagem e deslocamento. Crio também dentro da função a variável alfabeto e coloco as 25 letras do alfabeto, crio também a variável criptografada que irá ficar vazia para armazenar o resultado.

Faço um for que vai percorrer cada caracter da mensagem, crio a váriavel posicao e dentro dela utilizo alfabeto.index(), o index será responsável por descobrir a posição da letra no alfabeto.

Depois crio a variável nova_posicao que fará a soma da posição + deslocamento e o resultado será divido por 26 para voltar ao ínicio do alfabeto.

Após isso adiciono a nova letra na variável criptografada e uso o return para retornar o resultado.

Fora da função eu faço a entrada de dados para a mensagem pedindo para que o usuário escreva uma mensagem, também faço o mesmo para o deslocamento pedindo para que coloque o número para deslocamento.

Uso o print no final para imprimir como ficou a mensagem criptografada, novamente utilizo o format para preencher a chave com o resultado: ".format(criptografar(mensagem,deslocamento)))".

```python
def criptografar(mensagem, deslocamento):
  alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
  criptografada = ""
  for c in mensagem:
    posicao = alfabeto.index(c)
    nova_posicao = (posicao + deslocamento) % 26
    criptografada = criptografada + alfabeto[nova_posicao]
  return criptografada

mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite um número para deslocamento: "))
print("A mensagem CRIPTOGRAFADA ficou do seguinte modo: {}".format(criptografar(mensagem,deslocamento)))
```

```python
def criptografar(mensagem, deslocamento):
  alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
  criptografada = ""
  for c in mensagem:
    posicao = alfabeto.index(c)
    nova_posicao = (posicao + deslocamento) % 26
    criptografada = criptografada + alfabeto[nova_posicao]
  return criptografada

def descriptografar(mensagem, deslocamento):
  alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
  descriptografada = ""
  for c in mensagem:
    posicao = alfabeto.index(c)
    nova_posicao = (posicao - deslocamento) % 26
    descriptografada = descriptografada + alfabeto[nova_posicao]
  return descriptografada

mensagem = input("Digite uma mensagem para ser criptografada: ")
deslocamento = int(input("Digite um número de deslocamento para a mensagem: "))

cifra = criptografar(mensagem, deslocamento)
print("Mensagem CRIPTOGRAFADA ficaria da seguinte maneira: {}".format(cifra))

sem_cifra = descriptografar(cifra, deslocamento)
print("A mensagem DESCRIPTOGRAFADA era: {}".format(sem_cifra))
```

Crio uma variável para pegar os 9 primeiros dígitos cujo o nome será primeiros e crio uma segunda variável que irá pegar os números após os 9 primeiros, ou seja, o digito.

Crio um for com range(9) que irá repetir 9 vezes para cada número do cpf, eu crio uma variável chamada peso e dentro dela eu faço o 10 - n para conseguir os pesos e depois uso a soma para realizar a conta da multiplicação.

Depois utilizo o resto = soma % 11 para pegar o resto da divisão e faço o if para verificar se o resto é menor que 2. Caso seja, o primeiro dígito será 0, caso contrário faço 11 - resto para encontrar o primeiro dígito.

Depois adiciono o primeiro dígito aos 9 primeiros números e faço novamente o cálculo, porém dessa vez utilizando range(10) e os pesos começando em 11 para encontrar o segundo dígito.

Por fim, junto os dois dígitos calculados e comparo com os dois últimos números do CPF. Se forem iguais, o CPF será válido, caso contrário será inválido.

```python
def validar_cpf(cpf):
    cpf = cpf.replace(".", "")
    cpf = cpf.replace("-", "")

    primeiros = cpf[:9]
    digito = cpf[9:]

    soma = 0

    for n in range(9):
        peso = 10 - n
        soma = soma + int(primeiros[n]) * peso

    resto = soma % 11

    if resto < 2:
        primeiro_digito = 0
    else:
        primeiro_digito = 11 - resto

    segundos = primeiros + str(primeiro_digito)

    soma = 0

    for n in range(10):
        peso = 11 - n
        soma = soma + int(segundos[n]) * peso

    resto = soma % 11

    if resto < 2:
        segundo_digito = 0
    else:
        segundo_digito = 11 - resto

    calculado = str(primeiro_digito) + str(segundo_digito)

    if calculado == digito:
        print("O CPF é válido!")
    else:
        print("O CPF é inválido!")


cpf = input("Digite o seu CPF: ")
validar_cpf(cpf)
```

Por fim, crio a variável numero para armazenar a entrada de dados digitada pelo usuário e utilizo o print() para mostrar o resultado do fatorial do número digitado.

```python
def fatorial(n):
  if n == 1:
    return 1
  elif  n == 0:
    return 0
  else:
    return n * fatorial(n - 1)

numero = int(input("Digite um número para calcular o fatorial: "))
print("O fatorial de {} é: {}".format(numero, fatorial(numero)))
```

Por fim, utilizo outro if para verificar se a variavel primo continua como True. Se estiver, os números primos serão imprimidos e exibidos na tela para o usuário.

```python
numero = int(input("Digite um número para ver a sequência de primos até ele: "))
for numero in range(2, numero + 1):
  primo = True
  for divisor in range(2, numero):
    if numero % divisor == 0:
      primo = False
  if primo:
    print(numero, end=" ")
```

Exercício 46: Calcular a Média e Desvio Padrão

Neste exercício, o objetivo era calcular a média e o desvio padrão de uma lista de números.

Para isso começo importando a biblioteca math, que será utilizada para fazer a raiz quadrada. Depois crio a lista numeros com seguintes números: numeros = [10, 20, 30, 40, 50].

Em seguida crio a variável media, onde utilizo sum() para somar todos os números e len() para dividir pela quantidade de números.

Depois crio a variável soma começando com 0 e utilizo um for para percorrer os números. Dentro dele faço a diferença entre cada número e a média, elevo o resultado ao quadrado e adiciono na variável soma.

Depois crio a variável desvio e utilizo math.sqrt() para calcular a raiz quadrada da soma dividida pela quantidade de números.

Por fim utilizo o print() para mostrar a média e o desvio padrão.

```python
import math
numeros = [10, 20, 30, 40, 50]
media = sum(numeros) / len(numeros)
soma = 0
for n in numeros:
    soma += (n - media) ** 2
desvio = math.sqrt(soma / len(numeros))

print("Média:", media)
print("Desvio padrão:", desvio)
```

Exercício 48: Ordenação por inserção

Neste exercício, o objetivo era implementar o algoritmo de ordenação por inserção (Insertion Sort).

Para isso começo criando uma lista chamada numeros com alguns números. Depois utilizo um for com range(1, len(numeros)) para percorrer a lista começando pelo segundo número.

Dentro do for crio a variável atual para armazenar o número que será organizado e a variável j para pegar a posição anterior.

Depois utilizo o while para verificar se o número anterior é maior que o número atual. Se for, ele é movido uma posição para frente e o j diminui para continuar verificando os números anteriores.

Quando encontro a posição correta, coloco o número armazenado em atual nela. No fim utilizo o print() para mostrar a lista já ordenada.

```python
numeros = [5, 2, 8, 1]

for i in range(1, len(numeros)):
    atual = numeros[i]
    j = i - 1
    while j >= 0 and numeros[j] > atual:
        numeros[j + 1] = numeros[j]
        j -= 1
    numeros[j + 1] = atual
print("A lista ordenada ficaria da seguinte maneira: {}".format(numeros))
```
